System Health Check

```python
import psutil

cpu = psutil.cpu_percent()

memory = psutil.virtual_memory()

disk = psutil.disk_usage('C:')

print("CPU:", cpu)
print("RAM:", memory.percent)
print("Disk:", disk.percent)
```

Event Log Collection

```python
import subprocess

command = [
    "powershell",
    "Get-EventLog",
    "-LogName",
    "System",
    "-Newest",
    "20"
]

result = subprocess.run(
    command,
    capture_output=True,
    text=True
)

print(result.stdout)
```

```python
import os

os.makedirs("reports", exist_ok=True)

with open("reports/logs.txt", "w") as f:
    f.write(result.stdout)

print("Logs saved successfully")
```

```python
import pandas as pd
from datetime import datetime

health_report = {
    "Date": [datetime.now()],
    "CPU_Usage": [15.5],
    "RAM_Usage": [85.9],
    "Disk_Usage": [54.1]
}

df = pd.DataFrame(health_report)

print(df)
```

```python
df.to_csv(
    r"C:\Users\User\Documents\system_health_report.csv",
    index=False
)

print("Report saved successfully")
```

```python
def check_status(value):
    if value < 70:
        return "Healthy"
    elif value < 90:
        return "Warning"
    else:
        return "Critical"

df["CPU_Status"] = df["CPU_Usage"].apply(check_status)
df["RAM_Status"] = df["RAM_Usage"].apply(check_status)
df["Disk_Status"] = df["Disk_Usage"].apply(check_status)

print(df)
```
